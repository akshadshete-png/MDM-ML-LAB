```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("retail_sales_dataset.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df.drop_duplicates()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='Gender', data=df)
plt.title("Gender Distribution")
plt.show()
```

```python
gender = df['Gender'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(gender,
        labels=gender.index,
        autopct='%1.1f%%',
        startangle=90)
plt.title("Gender Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Age'], bins=10, kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.kdeplot(df['Total Amount'], fill=True)
plt.title("Distribution of Total Amount")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x=df['Total Amount'])
plt.title("Box Plot of Total Amount")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.violinplot(x='Gender', y='Total Amount', data=df)
plt.title("Violin Plot of Total Amount by Gender")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(x='Age', y='Total Amount', data=df, alpha=0.6)
plt.title("Scatter Plot of Total Amount vs Age")
plt.xlabel("Age")
plt.ylabel("Total Amount")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(y='Total Amount', data=df)
plt.title("Box Plot of Total Amount")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
# Select only numeric columns for correlation
corr = df.select_dtypes(include=['number']).corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Correlation Heatmap of Numeric Variables")
plt.show()
```

```python
sns.pairplot(df.select_dtypes(include='number'))
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.stripplot(x='Gender', y='Total Amount', data=df, alpha=0.5)
plt.title("Strip Plot of Total Amount by Gender")
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# Scale the numerical features
columns_to_scale = ['Age', 'Price per Unit', 'Quantity', 'Total Amount']
df[columns_to_scale] = scaler.fit_transform(df[columns_to_scale])

print(df.head())
```

```python
from sklearn.preprocessing import MinMaxScaler

# Instantiate the MinMaxScaler
min_max_scaler = MinMaxScaler()

# Normalize the numerical columns
cols = ['Age', 'Price per Unit', 'Quantity', 'Total Amount']
df[cols] = min_max_scaler.fit_transform(df[cols])

print(df.head())
```
